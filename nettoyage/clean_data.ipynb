```python
import numpy as np
import pandas as pd

df=pd.read_excel("XPON_QUALIF(1).xlsx")
df.head()
```

```python
df.info()
```

```python
isnull = df.isnull()

isnull_count = isnull.sum()

print(isnull_count)

total_isnull = isnull.sum().sum()
print("totals null : ", total_isnull)
```

```python
df.dropna(inplace=True)

df
```

```python
doublons = df.duplicated()

doublons_count = doublons.sum()

print(doublons_count)
print(df[doublons])
```

```python
df = df.drop_duplicates()
```

```python
def etat_valeur(v, low, high):
    if v < low:
        return "faible"
    elif v <= high:
        return "bonne"
    else:
        return "elevee"
df["Etat_OLT_TX"] = df["OLT_POWER_TX"].apply(lambda x: etat_valeur(x, 2, 7))
df["Etat_OLT_RX"] = df["OLT_RX_ONT_POWER"].apply(lambda x: etat_valeur(x, -32, -12))
df["Etat_ONU_TX"] = df["ONU_POWER_TX"].apply(lambda x: etat_valeur(x, 2, 5))
df["Etat_ONU_RX"] = df["ONU_POWER_RX"].apply(lambda x: etat_valeur(x, -28, -7))
```

```python
def etat_global(row):
    if "faible" in row.values:
        return "anomalie"
    return "normal"

df["Anomalie_technique"] = df[
    ["Etat_OLT_TX","Etat_OLT_RX","Etat_ONU_TX","Etat_ONU_RX"]
].apply(etat_global, axis=1)
```

```python

df["Anomalie_admin"] = np.where(
    (df["F_SUSP"] == 1) & (df["ADMIN_STATUS"] == 1),
    "Anomalie",
    "Normal"
)

```

```python

df["Anomalie"] = np.where(
    (df["Anomalie_technique"] == "anomalie") | (df["Anomalie_admin"] == "anomalie"),
    "Anomalie",
    "Normal"
)

```

```python
df.to_excel("dataset_visu.xlsx",index=False)
```
