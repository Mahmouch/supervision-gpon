Le nettoyage des données, également appelé nettoyage ou épuration des données, consiste à détecter et à supprimer les erreurs et les incohérences des données afin d’améliorer leur qualité. Les problèmes de qualité des données sont présents dans les données uniques
collections, telles que des fichiers et des bases de données, par exemple, en raison de fautes d’orthographe lors de la saisie de données, d’informations manquantes ou autres données invalides.

Enrichissement du dataset selon les règles

```python
import pandas as pd
import numpy as np

df_original = pd.read_excel("XPON_QUALIF.xlsx")
n_original = len(df_original)

n_to_add = np.random.randint(15000, 20001)

ID_LIGNE_VALUES = df_original['ID_LIGNE'].unique()
ID_CARTE_OLT_VALUES = df_original['ID_CARTE_OLT'].unique()
NO_PORT_VALUES = list(range(1, 17))
ONU_ID_VALUES = list(range(1, 101))

def generate_synthetic_rows(n):
    rows = []

    for _ in range(n):

        category = np.random.choice(
            ['faible', 'bon', 'eleve'],
            p=[0.15, 0.7, 0.15]
        )


        if category == 'faible':
            olt_tx = np.random.uniform(0, 2)
        elif category == 'bon':
            olt_tx = np.random.uniform(2, 7)
        else:
            olt_tx = np.random.uniform(7, 12)


        if category == 'faible':
            olt_rx = np.random.uniform(-40, -32)
        elif category == 'bon':
            olt_rx = np.random.uniform(-32, -12)
        else:
            olt_rx = np.random.uniform(-12, -5)


        if category == 'faible':
            onu_tx = np.random.uniform(0, 2)
        elif category == 'bon':
            onu_tx = np.random.uniform(2, 5)
        else:
            onu_tx = np.random.uniform(5, 10)

        if category == 'faible':
            onu_rx = np.random.uniform(-35, -28)
        elif category == 'bon':
            onu_rx = np.random.uniform(-28, -7)
        else:
            onu_rx = np.random.uniform(-7, 0)

        if np.random.random() < 0.10:
            f_susp = 1
            admin_status = 0 if np.random.random() < 0.8 else 1
        else:
            f_susp = 0
            admin_status = np.random.choice([0, 1])

        oper_status = 0 if (f_susp == 1 or admin_status == 0) else 1

        if np.random.random() < 0.05:
            olt_tx += np.random.uniform(-5, 5)
            onu_rx += np.random.uniform(-10, 10)

        rows.append({
            'ID_LIGNE': np.random.choice(ID_LIGNE_VALUES),
            'ID_CARTE_OLT': np.random.choice(ID_CARTE_OLT_VALUES),
            'NO_PORT': np.random.choice(NO_PORT_VALUES),
            'ONU_ID': np.random.choice(ONU_ID_VALUES),
            'OLT_POWER_TX': round(olt_tx, 2),
            'OLT_RX_ONT_POWER': round(olt_rx, 2),
            'ONU_POWER_TX': round(onu_tx, 2),
            'ONU_POWER_RX': round(onu_rx, 2),
            'F_SUSP': f_susp,
            'ADMIN_STATUS': admin_status,
            'OPER_STATUS': oper_status
        })

    return pd.DataFrame(rows)

df_synthetic = generate_synthetic_rows(n_to_add)

df_final = pd.concat([df_original, df_synthetic], ignore_index=True)

df_final.to_excel("XPON_QUALIF_augmented.xlsx", index=False)

print(f"Dataset original : {n_original} lignes")
print(f"Lignes ajoutées  : {n_to_add} lignes")
print(f"Dataset final   : {len(df_final)} lignes")
```

Chaque colonne technique est analysée indépendamment

De nouvelles variables dérivées sont créées

Le dataset devient interprétable par un humain (analyse métier, visualisation)

Cette étape constitue une préparation essentielle avant la détection d’anomalies

```python
def etat_valeur(v, low, high):
    if v < low:
        return "faible"
    elif v <= high:
        return "bonne"
    else:
        return "elevee"
df["Etat_OLT_TX"] = df["OLT_POWER_TX"].apply(lambda x: etat_valeur(x, 2, 7))
df["Etat_OLT_RX"] = df["OLT_RX_ONT_POWER"].apply(lambda x: etat_valeur(x, -32, -12))
df["Etat_ONU_TX"] = df["ONU_POWER_TX"].apply(lambda x: etat_valeur(x, 2, 5))
df["Etat_ONU_RX"] = df["ONU_POWER_RX"].apply(lambda x: etat_valeur(x, -28, -7))
```

Cette fonction applique une logique OR

Elle permet de détecter des anomalies partielles, fréquentes en réseau

Elle reflète une approche réaliste de la supervision proactive

```python
def etat_global(row):
    if "faible" in row.values:
        return "anomalie"
    return "normal"

df["Anomalie_technique"] = df[
    ["Etat_OLT_TX","Etat_OLT_RX","Etat_ONU_TX","Etat_ONU_RX"]
].apply(etat_global, axis=1)
```

ajouter une nouvelle colonne pour detecter l'anomalie administrative selon:

F_SUSP = 1 → ligne suspendue

ADMIN_STATUS = 1 → action administrative active

```python

df["Anomalie_admin"] = np.where(
    (df["F_SUSP"] == 1) & (df["ADMIN_STATUS"] == 1),
    "Anomalie",
    "Normal"
)

```

ajouter une nouvelle colonne pour la detection d'anomalies globales

```python

df["Anomalie"] = np.where(
    (df["Anomalie_technique"] == "Anomalie") | (df["Anomalie_admin"] == "Anomalie"),
    "Anomalie",
    "Normal"
)

df
```

exporter la nouvelle dataset sous format xlsx qui est spécifique au la partie de visualisation

```python
df.to_excel("dataset_vis.xlsx",index=False)
```
